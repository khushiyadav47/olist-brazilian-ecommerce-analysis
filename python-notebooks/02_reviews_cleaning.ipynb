```python
import pandas as pd
df = pd.read_csv('olist_order_reviews_dataset.csv')
df.head()
```

```python
df['review_score'].value_counts()
```

```python
df['review_answer_timestamp'] = pd.to_datetime(df['review_answer_timestamp'])
df['review_creation_date'] = pd.to_datetime(df['review_creation_date'])
df.dtypes
```

```python
df['review_answer_timestamp'] = df['review_answer_timestamp'].dt.normalize()
df['review_creation_date'] = df['review_creation_date'].dt.normalize()
```

```python
df.head()
```

```python
df[df['order_id'].duplicated(keep = False)]
```

```python
#sorting the review creation column according to its date
df = df.sort_values('review_creation_date')
df.head(10)
```

```python
df = df.drop_duplicates(subset = 'order_id', keep = 'last')
df.shape
```

```python
df = df.drop(columns = ['review_comment_message','review_comment_title'])
df.dtypes
```

```python
df.to_csv('reviews_clean.csv', index = False)

from google.colab import files
files.download('reviews_clean.csv')
```
