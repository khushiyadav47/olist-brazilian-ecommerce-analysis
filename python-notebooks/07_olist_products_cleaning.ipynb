```python
df1 = df.merge(df_translation, on ='product_category_name',how = 'left')
```

```python
df1.head()
```

```python
df1.shape
```

```python
df1.isnull().sum()
```

```python
df1['product_category_name_english'] = df1['product_category_name_english'].fillna('unknown')
df1.isnull().sum()
```

```python
df1 = df1.drop(columns = ['product_category_name','product_name_lenght','product_description_lenght','product_photos_qty','product_weight_g','product_length_cm','product_height_cm','product_width_cm'])
df1.info()
```

```python
df1 = df1.rename(columns = {'product_category_name_english':'product_category'})
df1.info()
```

```python
df1.to_csv('products_clean.csv', index = False)

from google.colab import files
files.download('products_clean.csv')
```
