```python
import pandas as pd
df = pd.read_csv('olist_sellers_dataset.csv')
df.head()
```

```python
df['seller_city'] = df['seller_city'].str.lower()
df['seller_city'] = df['seller_city'].str.strip()
```

```python
df['seller_city'] = df['seller_city'].str.replace('são paulo', 'sao paulo')
```

```python
df['seller_city'] = df['seller_city'].str.split(r'[/\-,]').str[0].str.strip()
```
