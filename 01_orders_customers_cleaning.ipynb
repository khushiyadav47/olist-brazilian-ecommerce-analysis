```python
#order_status k amoung kitne cols me null values h ye check krna
olist_orders.groupby('order_status')[['order_approved_at',
                                   'order_delivered_carrier_date',
                                   'order_delivered_customer_date']].apply(lambda d : d.isnull().sum())
```

```python
orders_clean = olist_orders.copy()
orders_clean.head()
```

```python
#converting the object datatype to date datatype from object datatype
orders_clean['order_delivered_carrier_date'] = pd.to_datetime(orders_clean['order_delivered_carrier_date'])
orders_clean['order_delivered_customer_date'] = pd.to_datetime(orders_clean['order_delivered_customer_date'])
orders_clean['order_estimated_delivery_date'] = pd.to_datetime(orders_clean['order_estimated_delivery_date'])
orders_clean['order_approved_at'] = pd.to_datetime(orders_clean['order_approved_at'])
orders_clean.head(10)
```

```python
orders_clean.dtypes
```

```python
#now converthing the format of date and time in the col of dates to only date by normalize() function
orders_clean['order_approved_at'] = orders_clean['order_approved_at'].dt.normalize()
orders_clean['order_delivered_carrier_date'] = orders_clean['order_delivered_carrier_date'].dt.normalize()
orders_clean['order_delivered_customer_date'] = orders_clean['order_delivered_customer_date'].dt.normalize()
orders_clean['order_estimated_delivery_date'] = orders_clean['order_estimated_delivery_date'].dt.normalize()
```

```python
orders_clean[['order_approved_at','order_delivered_carrier_date','order_estimated_delivery_date','order_delivered_customer_date']].head()
```

```python
#creating a new col delay_time which will contain the diff of actual delivery date and estimated delivery date
orders_clean['delay_time'] = (orders_clean['order_delivered_customer_date'] - orders_clean['order_estimated_delivery_date']).dt.days
```

```python
#now converting the delay time into categories
import numpy as np

conditions = [
    orders_clean['delay_time'] < 0,
    orders_clean['delay_time'] == 0,
    ((orders_clean['delay_time'] > 0) & (orders_clean['delay_time'] <= 3)),
    orders_clean['delay_time'] > 3
]

labels = ['early','on-time','slightly-late','late']

orders_clean['delivery_status'] = np.select(conditions, labels, default = None)
```

```python
orders_clean['delivery_status'].value_counts()
```

```python
orders_clean[(orders_clean['order_status'] != 'delivered') & (orders_clean['delay_time'].notnull())].shape
```

```python
orders_clean = orders_clean.drop('delay_time',axis = 1)
```

```python
orders_clean.head()
```

```python
from google.colab import files
orders_clean.to_csv('olist_orders_clean.csv', index = False)
files.download('olist_orders_clean.csv')
```
